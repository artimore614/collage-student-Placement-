```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("college_student_placement_dataset.csv")

df.head()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Placement", data=df)
plt.title("Placement Distribution")
plt.xlabel("Placement Status")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    x="Internship_Experience",
    hue="Placement",
    data=df
)

plt.title("Internship Experience vs Placement")
plt.xlabel("Internship Experience")
plt.ylabel("Number of Students")
plt.legend(title="Placement")
plt.show()
```

### This graph compares placement status between students with and without internship experience. It helps us understand the relationship between internship experience and placement.

```python
plt.figure(figsize=(8,5))

sns.boxplot(x="Placement", y="CGPA", data=df)

plt.title("CGPA vs Placement")
plt.xlabel("Placement Status")
plt.ylabel("CGPA")

plt.show()
```

### “Students with different placement statuses can be compared based on their CGPA. This visualization helps identify whether CGPA has a relationship with placement.”

```python
plt.figure(figsize=(8,5))

sns.boxplot(x="Placement", y="Academic_Performance", data=df)

plt.title("Academic Performance vs Placement")
plt.xlabel("Placement Status")
plt.ylabel("Academic Performance")

plt.show()
```

### “The graph compares the academic performance scores of placed and non-placed students and helps analyze the relationship between academic performance and placement status.

```python
plt.figure(figsize=(8,5))

sns.boxplot(x="Placement", y="Communication_Skills", data=df)

plt.title("Communication Skills vs Placement")
plt.xlabel("Placement Status")
plt.ylabel("Communication Skills Score")

plt.show()
```

### “This graph compares the communication skills scores of placed and non-placed students to analyze whether communication skills influence placement.”

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="Projects_Completed",
    hue="Placement",
    data=df
)

plt.title("Projects Completed vs Placement")
plt.xlabel("Number of Projects Completed")
plt.ylabel("Number of Students")
plt.legend(title="Placement")

plt.show()
```

### “This graph compares the placement status of students based on the number of projects they completed. It helps analyze whether completing more projects is associated with better placement outcomes.”

```python
plt.figure(figsize=(10,7))

# Convert categorical columns to numeric
df_heatmap = df.copy()

df_heatmap["Internship_Experience"] = df_heatmap["Internship_Experience"].map({
    "No": 0,
    "Yes": 1
})

df_heatmap["Placement"] = df_heatmap["Placement"].map({
    "No": 0,
    "Yes": 1
})

# Calculate correlation
correlation = df_heatmap.corr(numeric_only=True)

# Create heatmap
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```

### “The correlation heatmap shows the relationship between numerical features and placement status. Features with stronger positive or negative correlation may have a greater influence on placement prediction.”
